## Materialize the selected response

This page reconstructs the exact request from lesson 2, then gives it to
the same `CircuitRun`. The `DirectSolveSpec` asks for the complete
selected-view S/Y/Z response; the named trace is only a projection.
`run.solve()` returns a typed `DirectSolveResult` rather than storing a
current result on the Run.

```python
from fixtures.primitive_resonator import build_primitive_resonator
from scnsim import (
    CircuitRun,
    DirectSolveSpec,
    ReductionPipeline,
    SParameterTrace,
    units as u,
)

fixture = build_primitive_resonator()
run = CircuitRun(plan=fixture.plan, workspace="workspaces/primitive-course")
view = run.original.reduce(ReductionPipeline().retain("signal_in"))
spec = DirectSolveSpec(
    frequencies=[5.5, 6.0, 6.5, 7.0] * u.GHz,
    traces=(
        SParameterTrace(
            id="reflection",
            input_port="signal_in",
            input_mode=(),
            output_port="signal_in",
            output_mode=(),
        ),
    ),
)
direct = run.solve(view, spec)
```

## Present the existing Result

`direct.s`, `direct.y`, and `direct.z` are parallel matrix-family
surfaces. `show()` only presents materialized data; it never solves
again.

```python
direct.traces["reflection"].show(magnitude="db")
direct.s.view.coordinates
direct.y.view.matrix
direct.z.view.matrix
```
