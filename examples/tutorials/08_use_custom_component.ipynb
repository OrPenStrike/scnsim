## Compose the reusable subsystem

The package catalog owns how the LC is built; the outer Plan still owns
where its one public terminal connects. Built-in `components` and the
custom catalog both return exact-identity `ComponentInstance`s, so
`plan.add()` treats them the same way.

```python
from fixtures import resonator_library
from scnsim import CircuitDiagramSpec, CircuitPlan, components, units as u

plan = CircuitPlan(id="composite_resonator")
coupling_cap = plan.add(
    components.capacitor(id="coupling_cap", capacitance=6.0 * u.fF)
)
resonator = plan.add(
    resonator_library.components.parallel_linear_lc_resonator(
        id="resonator",
        capacitance=110.0 * u.fF,
        inductance=5.8 * u.nH,
    )
)

signal_boundary = plan.net(coupling_cap.pin("terminal_1"))
plan.net(
    coupling_cap.pin("terminal_2"),
    resonator.pin("terminal"),
    id="resonator_node",
)
plan.add_port(
    id="signal_in",
    at=signal_boundary,
    role="terminated",
    reference_impedance=50.0 * u.ohm,
)
```

## Inspect the abstraction you authored

The authoring diagram preserves the custom component boundary. A
compiled diagram may expand it for audit, but neither projection
replaces the Plan.

```python
diagram = plan.render_schematic(
    CircuitDiagramSpec(representation="authoring", show_parameter_values=True)
)
diagram.show()
```
