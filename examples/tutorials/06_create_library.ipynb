## Name the catalog your project owns

`scnsim.components` is the built-in catalog. A team that owns reusable
project components should publish a different, clearly named Python
module rather than register factories in global Notebook state.
`Library` is the base class for that custom catalog; it intentionally
does not inherit the built-in factories.

SCNSim derives catalog identity from the concrete subclass and its
installed or source provenance. The author supplies no manual identity
string.

```python
from scnsim import Library


class ResonatorLibrary(Library):
    """Project-specific reusable resonator components."""


components = ResonatorLibrary()
```

## Inspect before adding a factory

Python’s native documentation tools are enough to inspect this empty
catalog. It has a stable type and docstring but no custom component
factory yet. The next lesson moves this pattern into the canonical
project module and defines the first factory there.

```python
help(ResonatorLibrary)
help(components)
```
