# 2 — Prepare a Direct request

> **In 3–5 minutes**
>
> **Learn:** separate a sealed Run, lazy network view, and response
> request. **Run:** build one `DirectSolveSpec` without solving it.
> **Inspect:** deterministic preflight evidence. **Status:**
> `CONVERGING` scaffold; compute wait excluded.

## Choose the analysis view and request

The adjacent [construction fixture](fixtures/primitive_resonator.py)
only repeats lesson 1’s Plan and handles. `CircuitRun` owns execution,
while `run.original` and `retain()` describe a lazy [network-view
lineage](../../docs/concepts/compilation-coordinates-and-network-views.qmd#network-view-lineage).
`SParameterTrace` names one ordered matrix projection, while
`DirectSolveSpec` requests the complete selected-view S/Y/Z response on
an explicit frequency grid.

```python
from fixtures.primitive_resonator import build_primitive_resonator
from scnsim import (
    CircuitRun,
    DirectSolveSpec,
    ReductionPipeline,
    SParameterTrace,
    units as u,
)

fixture = build_primitive_resonator()
run = CircuitRun(plan=fixture.plan, workspace="workspaces/primitive-course")
response_view = run.original.reduce(
    ReductionPipeline().retain("signal_in")
)
reflection = SParameterTrace(
    id="reflection",
    input_port="signal_in",
    input_mode=(),
    output_port="signal_in",
    output_mode=(),
)
direct_spec = DirectSolveSpec(
    frequencies=[5.5, 6.0, 6.5, 7.0] * u.GHz,
    traces=(reflection,),
)
```

## Inspect without solving

`run.explain()` deterministically compiles the request boundary for
review. It returns inspection evidence, writes no workspace artifact,
and creates no numerical Result.

```python
explanation = run.explain(response_view, direct_spec)
explanation.show()
```
